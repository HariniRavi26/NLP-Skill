```python
%pip install -q datasets

import re
from collections import Counter, defaultdict
from datasets import load_dataset

dataset = load_dataset("Salesforce/wikitext", "wikitext-2-raw-v1")

text = "\n".join(dataset["train"]["text"])

text = text.lower()
text = re.sub(r"[^a-zA-Z0-9\s']", " ", text)
text = re.sub(r"\s+", " ", text).strip()

tokens = text.split()

unigram_counts = Counter(tokens)

bigram_counts = Counter(
    zip(tokens, tokens[1:])
)

trigram_counts = Counter(
    zip(tokens, tokens[1:], tokens[2:])
)

bigram_next_words = defaultdict(Counter)

for (w1, w2), count in bigram_counts.items():
    bigram_next_words[w1][w2] = count

trigram_next_words = defaultdict(Counter)

for (w1, w2, w3), count in trigram_counts.items():
    trigram_next_words[(w1, w2)][w3] = count

def predict_next_words(sentence, top_n=5):
    words = sentence.lower().split()

    if len(words) >= 2:
        previous_two = tuple(words[-2:])
        candidates = trigram_next_words.get(previous_two, {})

        if candidates:
            total = sum(candidates.values())

            predictions = [
                (word, count / total)
                for word, count in candidates.items()
            ]

            predictions.sort(
                key=lambda x: x[1],
                reverse=True
            )

            return predictions[:top_n]

    if len(words) >= 1:
        previous_word = words[-1]
        candidates = bigram_next_words.get(previous_word, {})

        if candidates:
            total = sum(candidates.values())

            predictions = [
                (word, count / total)
                for word, count in candidates.items()
            ]

            predictions.sort(
                key=lambda x: x[1],
                reverse=True
            )

            return predictions[:top_n]

    return []

def show_predictions(sentence, top_n=5):
    predictions = predict_next_words(sentence, top_n)

    print("\nInput:", sentence)
    print("\nTop predicted next words:")

    if not predictions:
        print("No prediction found.")
        return

    for i, (word, probability) in enumerate(predictions, 1):
        print(f"{i}. {word} - {probability:.3f}")

print("WikiText-2 loaded successfully")
print("Total tokens:", len(tokens))
print("Unique words:", len(unigram_counts))

while True:
    sentence = input("\nEnter a sentence or partial sentence (type 'exit' to stop): ")

    if sentence.lower().strip() == "exit":
        print("Word predictor stopped.")
        break

    if not sentence.strip():
        print("Please enter some words.")
        continue

    show_predictions(sentence, 5)
```
