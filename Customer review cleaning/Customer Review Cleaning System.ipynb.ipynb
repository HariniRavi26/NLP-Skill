```python
%pip install pandas nltk -q

import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

print("=" * 80)
print("CUSTOMER REVIEW CLEANING SYSTEM")
print("=" * 80)

df = pd.read_csv("IMDB Dataset.csv")

print("\nDataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())

print("\nFirst 5 original reviews:")
display(df[['review', 'sentiment']].head())

print("\nMissing values:")
print(df.isnull().sum())

duplicate_count = df['review'].duplicated().sum()
print("\nNumber of duplicate reviews:", duplicate_count)

df = df.dropna(subset=['review'])
df = df.drop_duplicates(subset=['review'])

print("\nRemaining reviews:", len(df))

df['cleaned_review'] = df['review']

df['cleaned_review'] = df['cleaned_review'].str.lower()

df['cleaned_review'] = df['cleaned_review'].apply(
    lambda text: re.sub(r'<.*?>', ' ', text)
)

df['cleaned_review'] = df['cleaned_review'].apply(
    lambda text: re.sub(r'https?://\S+|www\.\S+', ' ', text)
)

df['cleaned_review'] = df['cleaned_review'].apply(
    lambda text: re.sub(r'\d+', ' ', text)
)

df['cleaned_review'] = df['cleaned_review'].apply(
    lambda text: re.sub(r'[^a-zA-Z\s]', ' ', text)
)

df['cleaned_review'] = df['cleaned_review'].apply(
    lambda text: re.sub(r'\s+', ' ', text).strip()
)

stop_words = set(stopwords.words('english'))

print("\nNumber of English stopwords:", len(stop_words))

df['cleaned_review'] = df['cleaned_review'].apply(
    word_tokenize
)

df['cleaned_review'] = df['cleaned_review'].apply(
    lambda words: [
        word for word in words
        if word not in stop_words
    ]
)

stemmer = PorterStemmer()

df['cleaned_review'] = df['cleaned_review'].apply(
    lambda words: [
        stemmer.stem(word)
        for word in words
    ]
)

df['cleaned_review'] = df['cleaned_review'].apply(
    lambda words: ' '.join(words)
)

pd.set_option('display.max_colwidth', 1000)
pd.set_option('display.width', 2000)

print("\n" + "=" * 80)
print("ORIGINAL REVIEW VS CLEANED REVIEW")
print("=" * 80)

display(
    df[['review', 'cleaned_review', 'sentiment']].head(20)
)

print("\n" + "=" * 80)
print("CLEANED REVIEW SENTENCES")
print("=" * 80)

for i, row in df.head(20).iterrows():
    print("\nReview", i + 1)
    print("Sentiment :", row['sentiment'])
    print("Cleaned   :", row['cleaned_review'])
    print("-" * 80)

print("\n" + "=" * 80)
print("CLEANED REVIEWS ONLY")
print("=" * 80)

display(
    df[['cleaned_review', 'sentiment']].head(20)
)

print("\nFinal dataset shape:", df.shape)
print("Final columns:", df.columns.tolist())

display(df.head(10))

df.to_csv("IMDB_Preprocessed.csv", index=False)

print("\n" + "=" * 80)
print("TASK COMPLETED SUCCESSFULLY")
print("=" * 80)
print("Output file: IMDB_Preprocessed.csv")
print("Total reviews:", len(df))
```
